```python
import cv2

image = cv2.imread("pages/page_1.png")

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

thresh = cv2.threshold(
    gray,
    180,
    255,
    cv2.THRESH_BINARY_INV
)[1]

contours, _ = cv2.findContours(
    thresh,
    cv2.RETR_TREE,
    cv2.CHAIN_APPROX_SIMPLE
)

cells = []

for cnt in contours:

    x, y, w, h = cv2.boundingRect(cnt)

    # фильтр размеров клетки
    if 20 < w < 60 and 20 < h < 60:

        ratio = w / float(h)

        # почти квадрат
        if 0.7 < ratio < 1.3:

            cells.append((x, y, w, h))

            cv2.rectangle(
                image,
                (x, y),
                (x + w, y + h),
                (0, 255, 0),
                2
            )

print("cells found:", len(cells))

cv2.imwrite("cells_detected.png", image)
```
